## Nivel 1: Columna de identidad [ride_id]

Este dato representa unicidad en este data set, debe ser único e irrepetible ya que cada uno indica el registro de cada viaje.

```python
# Número de identificadores únicos de viaje

df['ride_id'].nunique()
```

```python
# Longitud de los identificadores de viaje

df['ride_id'].astype('str').str.len().value_counts()
```

```python
# Tipos de datos en la columna 'ride_id'

df['ride_id'].apply(type).value_counts()
```

```python
# Tipos de datos en la columna 'ride_id'

df['ride_id'].dtype
```

```python
# Inspección visual de los primeros 15 valores en 'ride_id'

df['ride_id'].head(15)
```

```python
# Detección de posibles problemas en 'ride_id'

ide=df['ride_id'].astype('string')

extremos=ide != ide.str.strip() # espacios al inicio o final
medio=ide.str.contains(r"\S\s+\S", regex=True, na=False) # espacios en el medio
tabs=ide.str.contains(r"[\t\n\r]", regex=True, na=False) # tabulaciones o saltos de línea
caracteres_especiales=ide.str.contains(r"[^\w-]", regex=True, na=False) # caracteres no alfanuméricos ni guiones

extremos.sum(), medio.sum(), tabs.sum(), caracteres_especiales.sum()
```

## Nivel 1 completo

En el EDA del nivel 1 se encontro que la columna de ID esta limpia:

 - Sin nulos

 - Sin duplicados

 - Misma longitud

 - Sin espacios (bordes/medio)

 - Sin tabs / saltos invisibles

 - Sin caracteres especiales raros

## Nivel 2: Fechas 

Las fechas aportan un valor analítico muy importante porque revelan tendencias, estacionalidad y cohortes, sin embargo, suelen ser los datos más problemas en su formato.

Para este data set tenemos started_at y ended_at y haremos el EDA en dos momentos

- Data Profiling (Validación estructural de fechas): formato consistente, longitud, rango, nulos, duplicados esperables
- Business Rules Validation (Coherencia temporal): inicio ≤ fin, duración válida, detección de outliers
